```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("It works!")
```

```python
conc = np.logspace(-9, -4, 10)          # 1 nM to 100 µM
response = 100 / (1 + (1e-6 / conc))    # simple inhibition-style curve
plt.semilogx(conc, response, "o-")
plt.xlabel("Concentration (M)")
plt.ylabel("Response (%)")
plt.show()
```
